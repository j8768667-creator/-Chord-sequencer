```python
# 1. Install dependencies
!pip install gradio scipy numpy -q

import gradio as gr
import numpy as np
from scipy.io import wavfile

def paulstretch_mono(samplerate, smp, stretch, windowsize_seconds=0.25):
    """
    Core Paulstretch DSP algorithm for a 1D mono float32 array.
    Uses windowed STFT with randomized phase to dissolve transients into texture.
    """
    n_window = int(samplerate * windowsize_seconds)
    if n_window % 2 != 0:
        n_window += 1
    half_window = n_window // 2

    # Hann window for clean crossfading
    hanning_win = 0.5 - 0.5 * np.cos(np.linspace(0, 2 * np.pi, n_window, endpoint=False))

    # Step sizes (hop lengths)
    hop_in = int(half_window / stretch)
    if hop_in < 1:
        hop_in = 1
    hop_out = half_window

    # Pad audio ends so we don't truncate boundaries
    smp = np.pad(smp, (0, n_window), mode='constant')
    total_samples = len(smp)

    out_length = int(total_samples * stretch) + n_window
    output = np.zeros(out_length, dtype=np.float32)

    in_pos = 0
    out_pos = 0

    half_spec = half_window + 1

    while in_pos + n_window < total_samples:
        # 1. Extract grain window
        grain = smp[in_pos : in_pos + n_window] * hanning_win

        # 2. Real FFT
        spec = np.fft.rfft(grain)

        # 3. Discard original phase; apply uniform random phase to diffuse transients
        mag = np.abs(spec)
        random_phase = np.exp(1j * (np.random.uniform(0, 2 * np.pi, half_spec)))
        stretched_spec = mag * random_phase

        # 4. Inverse FFT back to time-domain
        resynth = np.fft.irfft(stretched_spec, n=n_window) * hanning_win

        # 5. Overlap-add into output buffer
        output[out_pos : out_pos + n_window] += resynth

        in_pos += hop_in
        out_pos += hop_out

    # Trim to valid audio range
    return output[:out_pos]

def process_paulstretch(input_audio, stretch_factor, window_size):
    if input_audio is None:
        return None, None

    sr, audio = input_audio

    # Convert integer PCM to float32 (-1.0 to +1.0)
    if audio.dtype == np.int16:
        audio = audio.astype(np.float32) / 32768.0
    elif audio.dtype == np.int32:
        audio = audio.astype(np.float32) / 2147483648.0
    elif audio.dtype == np.uint8:
        audio = (audio.astype(np.float32) - 128.0) / 128.0

    # Process stereo channels independently for a rich spatial image
    if audio.ndim == 1:
        out_audio = paulstretch_mono(sr, audio, stretch_factor, window_size)
    else:
        left = paulstretch_mono(sr, audio[:, 0], stretch_factor, window_size)
        right = paulstretch_mono(sr, audio[:, 1], stretch_factor, window_size)
        min_len = min(len(left), len(right))
        out_audio = np.column_stack((left[:min_len], right[:min_len]))

    # Peak-normalize to -0.5 dB to prevent clipping
    peak = np.max(np.abs(out_audio))
    if peak > 0:
        out_audio = out_audio / peak * 0.95

    # Save uncompressed 16-bit WAV
    output_filename = f"paulstretch_{int(stretch_factor)}x.wav"
    wavfile.write(output_filename, sr, (out_audio * 32767).astype(np.int16))

    return output_filename, output_filename

# --- Gradio Web UI Layout ---
with gr.Blocks(title="Paulstretch Ambient Engine") as demo:
    gr.Markdown("## 🌌 Paulstretch Ambient Drone & Time-Stretch Engine")
    gr.Markdown("Upload any short audio sample or musical phrase to dissolve it into an infinite, lush ambient texture.")

    with gr.Row():
        with gr.Column(scale=1):
            audio_input = gr.Audio(label="Upload Audio (WAV, MP3, etc.)", type="numpy")
            stretch_slider = gr.Slider(
                minimum=2.0, maximum=50.0, value=8.0, step=1.0,
                label="Stretch Factor (Multiplier)"
            )
            window_slider = gr.Slider(
                minimum=0.05, maximum=1.0, value=0.25, step=0.05,
                label="Window Size (Seconds) — Larger = Softer/Glassier"
            )
            process_btn = gr.Button("Dissolve & Stretch", variant="primary")

        with gr.Column(scale=1):
            audio_player = gr.Audio(label="In-Browser Ambient Preview", type="filepath")
            download_file = gr.File(label="Download High-Res WAV")

    process_btn.click(
        fn=process_paulstretch,
        inputs=[audio_input, stretch_slider, window_slider],
        outputs=[audio_player, download_file]
    )

demo.launch(share=True)
```
